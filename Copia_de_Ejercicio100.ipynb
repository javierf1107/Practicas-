## Instrucciones
Implementa la clase `VehiculoAutonomoV10` respetando los siguientes requerimientos:

1. **`__init__(self, nombre: str, bateria: float)`**:
   - Inicializa las variables de instancia: `nombre` (str), `bateria` (float), `registro_viajes` (list vacía) y `activo` (bool = True).

2. **`evaluar_condicion(self, limite: float, factor: float) -> float`**:
   - Usa operadores aritméticos, lógicos y condicionales (`if / elif / else`).
   - Si `bateria` >= limite y `activo` es True: retorna `bateria` * factor.
   - Elif `bateria` < limite y `bateria` > 0: retorna `bateria` + 10.0.
   - Else: retorna 0.0.

3. **`procesar_registros(self, datos: list) -> float`**:
   - Usa un ciclo `for` para recorrer `datos` (lista de floats/ints/None).
   - Ignora los elementos que sean `None`.
   - Acumula los valores válidos en `bateria` y agrega cada valor a `registro_viajes`.
   - Si `bateria` supera los 500.0, desactiva la entidad (`activo = False`).
   - Retorna el valor final de `bateria`.

4. **`acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int`**:
   - Usa un ciclo `while` para incrementar `bateria` en `paso` mientras `bateria` sea estrictamente menor que `objetivo`.
   - Retorna la cantidad de iteraciones/pasos realizados.


```python
# Escribe tu solución aquí
class VehiculoAutonomoV10:
    def __init__(self, nombre: str, bateria: float):
        self.nombre = nombre
        self.bateria = bateria
        self.registro_viajes = []
        self.activo = True

    def evaluar_condicion(self, limite: float, factor: float) -> float:
        if self.bateria >= limite and self.activo:
            return self.bateria * factor
        elif self.bateria < limite and self.bateria > 0:
            return self.bateria + 10.0
        else:
            return 0.0

    def procesar_registros(self, datos: list) -> float:
        for valor in datos:
            if valor is not None:
                self.bateria = self.bateria + valor
                self.registro_viajes.append(valor)

                if self.bateria > 500.0:
                    self.activo = False

        return self.bateria

    def acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int:
        pasos = 0
        while self.bateria < objetivo:
            self.bateria = self.bateria + paso
            pasos = pasos + 1
        return pasos
```

```python
# ==============================================================================
# MOTOR DE AUTOEVALUACIÓN (NO MODIFICAR ESTA CELDA)
# ==============================================================================
import unittest

class TestEjercicio(unittest.TestCase):
    def test_01_init_y_atributos(self):
        obj = VehiculoAutonomoV10("Test", 50.0)
        self.assertTrue(hasattr(obj, 'nombre'), "Falta el atributo 'nombre'.")
        self.assertTrue(hasattr(obj, 'bateria'), "Falta el atributo 'bateria'.")
        self.assertTrue(hasattr(obj, 'registro_viajes'), "Falta el atributo 'registro_viajes'.")
        self.assertTrue(hasattr(obj, 'activo'), "Falta el atributo 'activo'.")
        self.assertEqual(getattr(obj, 'activo'), True, "El atributo 'activo' debe ser True por defecto.")

    def test_02_evaluar_condicion(self):
        obj = VehiculoAutonomoV10("Test", 100.0)
        res1 = obj.evaluar_condicion(50.0, 1.5)
        self.assertEqual(res1, 150.0, "Fallo en la lógica del condicional/multiplicación.")

        obj2 = VehiculoAutonomoV10("Test2", 20.0)
        res2 = obj2.evaluar_condicion(50.0, 1.5)
        self.assertEqual(res2, 30.0, "Fallo en la rama elif de evaluar_condicion.")

    def test_03_ciclo_for_y_colecciones(self):
        obj = VehiculoAutonomoV10("Test", 10.0)
        res = obj.procesar_registros([5.0, None, 15.0, 20.0])
        self.assertEqual(res, 50.0, "Fallo en acumulación del ciclo for o filtrado de None.")
        self.assertEqual(len(getattr(obj, 'registro_viajes')), 3, "No se ignoraron correctamente los valores None.")

    def test_04_ciclo_while(self):
        obj = VehiculoAutonomoV10("Test", 10.0)
        pasos = obj.acumular_hasta_objetivo(30.0, 5.0)
        self.assertEqual(pasos, 4, "Fallo en la cantidad de iteraciones del ciclo while.")
        self.assertEqual(getattr(obj, 'bateria'), 30.0, "Fallo en el incremento dentro del ciclo while.")

suite = unittest.TestLoader().loadTestsFromTestCase(TestEjercicio)
runner = unittest.TextTestRunner(verbosity=0)
res = runner.run(suite)

total = res.testsRun
errores = len(res.failures) + len(res.errors)
nota = round(((total - errores) / total) * 10, 1)

print("\n" + "="*45)
print(f" RESULTADO FINAL DE EVALUACIÓN")
print("="*45)
print(f" NOTA OBTENIDA: {nota} / 10.0\n")

if errores > 0:
    print("DETALLE DE DETECCIONES DE ERROR:")
    for f in res.failures:
        print(f" ❌ {f[1]}")
    for e in res.errors:
        print(f" ⚠️ Error de ejecución: {e[1]}")
else:
    print(" 🎉 ¡Excelente! Lógica de POO, ciclos y condicionales correcta.")
```
